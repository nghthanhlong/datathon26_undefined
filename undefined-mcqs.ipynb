```python
import numpy as np
import pandas as pd

import os
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))

input_dir='datasets/'
```

For MCQs questions

```python
#Cau 1
df=pd.read_csv(input_dir+'orders.csv')
df['order_date']=pd.to_datetime(df['order_date'])
df=df.sort_values(['customer_id', 'order_date'])
df["gap_days"] = df.groupby("customer_id")["order_date"].diff().dt.days
median_gap=df['gap_days'].median()
print(median_gap)
```

```python
#Cau 2
df=pd.read_csv(input_dir+'products.csv')
df[['segment', 'price', 'cogs']]
df['margin']=(df['price']-df['cogs'])/df['price']
result = (
    df.groupby('segment')['margin'].mean().sort_values(ascending=False)
)
print(result.head)
```

```python
#Cau 3
df_products=pd.read_csv(input_dir+'products.csv')
df_returns=pd.read_csv(input_dir+'returns.csv')

df_merge=pd.merge(df_products, df_returns, on='product_id')

res = (
    df_merge[df_merge['category']=="Streetwear"]['return_reason'].value_counts()
)
print(res.head)
```

```python
#Cau 4
df=pd.read_csv(input_dir+'web_traffic.csv')
df[["traffic_source", "bounce_rate"]]
result=df.groupby("traffic_source")['bounce_rate'].mean()
print(result)
```

```python
#Cau 5
df = pd.read_csv(
    input_dir+'order_items.csv',
    dtype={'promo_id': 'string'}
)

print(df['promo_id'].notna().mean()*100)
```

```python
#Cau 6
df_customer = pd.read_csv(input_dir+'customers.csv')
df_order = pd.read_csv(input_dir+'orders.csv')

df_customer = df_customer[df_customer['age_group'].notna()]

order_per_cus = df_order.groupby('customer_id')['order_id'].count().reset_index()

df_merge = pd.merge(df_customer, order_per_cus, on='customer_id', how='left')

df_merge['order_id'] = df_merge['order_id'].fillna(0)

res = df_merge.groupby('age_group')['order_id'].mean().sort_values(ascending=False)

print(res)
```

```python
#Cau 7
df_geo=pd.read_csv(input_dir+'geography.csv')
df_st=pd.read_csv(input_dir+'sales.csv')
df_od=pd.read_csv(input_dir+'orders.csv')
df_pm=pd.read_csv(input_dir+'payments.csv')

df_merge_od_pm=pd.merge(df_od, df_pm, on='order_id')
df_merge3=pd.merge(df_merge_od_pm, df_geo, on='zip')
res=df_merge3.groupby('region')['payment_value'].sum().sort_values(ascending=False)
print(res)
```

```python
#Cau 8
df=pd.read_csv(input_dir+'orders.csv')
df_cancel=df[df['order_status']=='cancelled']
counts = df_cancel.groupby('payment_method')['order_id'].count()
print(counts)
```

```python
#Cau 9
df_odi=pd.read_csv(input_dir+'order_items.csv')
df_r=pd.read_csv(input_dir+'returns.csv')
df_p=pd.read_csv(input_dir+'products.csv')

df_r['has_return'] = 1

df = df_odi.merge(df_p[['product_id', 'size']], on='product_id', how='left')

df = df.merge(df_r[['order_id', 'product_id', 'has_return']].drop_duplicates(),
              on=['order_id', 'product_id'], how='left')

df['has_return'] = df['has_return'].fillna(0)

res = df.groupby('size').agg(
    total_order_item_rows=('order_id', 'count'),
    total_return_records=('has_return', 'sum')
)
res['return_rate'] = res['total_return_records'] / res['total_order_item_rows']
res = res.sort_values('return_rate', ascending=False)
print(res)
```

```python
#Cau 10
df=pd.read_csv(input_dir+'payments.csv')
od_pay=df.groupby('order_id').agg({
    'payment_value': 'sum',
    'installments': 'first'
}).reset_index()
res=od_pay.groupby('installments')['payment_value'].mean().sort_values(ascending=False)
print(res)
```
